```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import io
import requests

# --- SECTION 1: LOAD DATA DIRECTLY FROM NASA ---
print("Fetching data from NASA OMNIWeb... (This may take 30 seconds)")

# URL for the official OMNI2 High-Res (Hourly) Dataset
url = "https://spdf.gsfc.nasa.gov/pub/data/omni/low_res_omni/omni2_all_years.dat"

# Define the columns we need (Based on OMNI2 format specification)
# Indices are 0-based. 
# 0=Year, 1=Day, 2=Hour, 8=Field_Mag, 12=Bz, 23=Density, 24=Speed, 38=Kp
col_names = ['Year', 'DOY', 'Hour', 'B_Mag', 'Bz', 'Density', 'Speed', 'Kp']
col_indices = [0, 1, 2, 8, 16, 23, 24, 38] # Note: Col 16 is usually Bz(GSM) which is best for physics

# Read the fixed-width text file
# NASA uses 999.9 or 9999 as "Missing". We will handle this later.
df = pd.read_csv(
    url, 
    delim_whitespace=True, 
    header=None, 
    usecols=col_indices,
    names=col_names,
    engine='python'
)

# --- SECTION 2: CLEANING & PRE-PROCESSING ---
print("Cleaning data...")

# 1. Create a Datetime Index (Year + Day of Year + Hour)
df['Date'] = pd.to_datetime(df['Year'] * 100000 + df['DOY'] * 100 + df['Hour'], format='%Y%j%H')
df = df.set_index('Date')
df = df.drop(columns=['Year', 'DOY', 'Hour'])

# 2. Filter for your Project Timeline (2015-2024)
df = df['2015-01-01':]

# 3. Handle NASA's "Missing Value" codes (999.9, 9999, etc.)
# We replace anything significantly large with NaN
df = df.replace([999.9, 9999.0, 9999, 99.99], np.nan)

# 4. Interpolate (Space weather is continuous; forward fill small gaps)
df = df.interpolate(method='time')

# 5. Fix Kp Index (NASA stores Kp=30 as 3.0. We convert to 0-9 scale)
df['Kp'] = df['Kp'] / 10.0

# 6. Create the Target Label (Storm vs. Quiet)
# A "Major Storm" is typically Kp >= 5.
df['Is_Storm'] = (df['Kp'] >= 5).astype(int)

# --- SECTION 3: FEATURE ENGINEERING ("ABOVE AND BEYOND") ---
# The "Newell Coupling" Approximation: E = v * Bz (simplified for now)
# Real formula: v^(4/3) * B^(2/3) * sin(theta)^8/3. 
# For this plot, we check the simple "Coupling" of Speed vs. Kp
df['Coupling_Simple'] = df['Speed'] * df['B_Mag']

print(f"Dataset Ready! Loaded {len(df)} hourly records.")
print(f"Number of Storm Hours found: {df['Is_Storm'].sum()}")

# --- SECTION 4: THE 'HYSTERESIS' PLOT ---
plt.figure(figsize=(10, 6))

# Plot Density of Quiet vs Storm data
sns.kdeplot(x=df[df['Is_Storm']==0]['Speed'], y=df[df['Is_Storm']==0]['Kp'], 
            cmap="Blues", fill=True, alpha=0.5, label='Quiet Time')
sns.kdeplot(x=df[df['Is_Storm']==1]['Speed'], y=df[df['Is_Storm']==1]['Kp'], 
            cmap="Reds", fill=True, alpha=0.6, label='Storm Event')

plt.title('The "Heartbeat" of a Solar Storm: Wind Speed vs. Kp Index', fontsize=14)
plt.xlabel('Solar Wind Speed (km/s)')
plt.ylabel('Geomagnetic Disturbance (Kp Index)')
plt.grid(True, alpha=0.3)
plt.legend(['Quiet (Kp < 5)', 'Storm (Kp >= 5)'])

plt.show()

# Display first few rows to confirm validity
df.head()
```

The Physics: The "Blue Blob" is the quiet, fast(er) solar wind. The "Red Blob" is the storm. The area where they meet (low speed, high density) often corresponds to Solar Wind Stagnation.

The "Cold Ion" Connection: When the solar wind is slow and dense (the bottom-right of your plot), plasma can "leak" into the Earth's magnetosphere more easily through Kelvin-Helmholtz instabilities. This forms a reservoir of "cold" ions that can later be energized during a storm.

```python
df.info()
# Create the "Newell Coupling" feature (The 'Cheat Code' for high accuracy)
# Formula approximation: Speed * (Bz if Bz < 0 else 0)
# This captures that energy only transfers when the magnetic field points SOUTH.

# 1. Create a "Southward Field" column (Keep negative Bz, set positive to 0)
df['Bz_South'] = df['Bz'].apply(lambda x: abs(x) if x < 0 else 0)

# 2. Calculate the Energy Coupling (Speed * Southward Field)
df['coupling_energy'] = df['Speed'] * df['Bz_South']

# 3. Check the correlation (It should be much higher than Speed alone!)
print("Correlation with Storms:")
print(df[['Speed', 'Bz', 'coupling_energy', 'Is_Storm']].corr()['Is_Storm'])
```

```python
# --- SECTION: FEATURE ENGINEERING (The "Secret Sauce") ---

print("Engineering Physics Features...")

# 1. Create 'Bz_South': The magnetic gatekeeper
# Physics: Energy only enters if Bz is South (Negative). If North (Positive), the door is closed.
df['Bz_South'] = df['Bz'].apply(lambda x: abs(x) if x < 0 else 0)

# 2. Calculate 'Coupling_Energy' (The Newell approximation)
# This represents the "Power" hitting the magnetosphere.
# Formula: Speed * Southward_Magnetic_Field
df['Coupling_Energy'] = df['Speed'] * df['Bz_South']

# 3. Create Lag Features (To predict 1 hour ahead)
# We shift the input data FORWARD by 1 hour, so row 't' contains data from 't-1'
df['Speed_Lag1'] = df['Speed'].shift(1)
df['Coupling_Lag1'] = df['Coupling_Energy'].shift(1)

# Drop the first row (which now has NaNs due to shifting)
df = df.dropna()

# --- VALIDATION: CORRELATION CHECK ---
# Does our new "Coupling" feature correlate better with Storms than raw Speed?
correlations = df[['Speed', 'Bz', 'Coupling_Energy', 'Is_Storm']].corr()['Is_Storm'].sort_values(ascending=False)

print("\n--- HYPOTHESIS TEST RESULTS ---")
print("Correlation with Storm Event (Is_Storm):")
print(correlations)

if correlations['Coupling_Energy'] > correlations['Speed']:
    print("\nSUCCESS: Your engineered physics feature beats raw data!")
else:
    print("\nNOTE: Raw speed is still dominant. We may need interaction trees.")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, f1_score, confusion_matrix

print("Preparing Data for Modeling...")

# 1. Define Features (X) and Target (y)
# We use the raw features + your WINNING 'Coupling_Energy' feature
features = ['Speed', 'Density', 'Bz', 'B_Mag', 'Coupling_Energy']
X = df[features]
y = df['Is_Storm']

# 2. STRICT TEMPORAL SPLIT (Crucial for validity!)
# We cannot use random shuffle. We must cut by time.
# Train on first 80%, Test on last 20%
split_point = int(len(df) * 0.8)

X_train, X_test = X.iloc[:split_point], X.iloc[split_point:]
y_train, y_test = y.iloc[:split_point], y.iloc[split_point:]

print(f"Training on {len(X_train)} hours. Testing on {len(X_test)} hours.")

# --- MODEL 1: THE BASELINE (Logistic Regression) ---
print("\nRunning Baseline (Logistic Regression)...")
lr = LogisticRegression(class_weight='balanced', max_iter=1000)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
f1_lr = f1_score(y_test, y_pred_lr)
print(f"Baseline F1 Score: {f1_lr:.3f}")

# --- MODEL 2: THE CHAMPION (Random Forest) ---
print("Running Champion Model (Random Forest)...")
rf = RandomForestClassifier(n_estimators=100, max_depth=10, class_weight='balanced', random_state=42)
rf.fit(X_train, y_train)
y_pred_rf = rf.predict(X_test)
f1_rf = f1_score(y_test, y_pred_rf)

print(f"Random Forest F1 Score: {f1_rf:.3f}")

# --- THE VERDICT ---
print("\n--- FINAL SCORECARD ---")
print(classification_report(y_test, y_pred_rf, target_names=['Quiet', 'Storm']))

if f1_rf > f1_lr:
    print(f"VICTORY: Random Forest beat Baseline by {((f1_rf - f1_lr)/f1_lr)*100:.1f}%!")
else:
    print("Hmm, closer than expected. We might need more trees.")

# Check Feature Importance (To prove 'Coupling' was the MVP)
import pandas as pd
importance = pd.Series(rf.feature_importances_, index=features).sort_values(ascending=False)
print("\nFeature Importance (What the model actually learned):")
print(importance)
```
